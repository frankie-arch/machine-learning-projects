```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
```

```python
columns = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety",
    "class"
]

car = pd.read_csv("car+evaluation/car.data", names=columns)
car
```

```python
car.info()
```

```python
for column in car.columns:
    print(car[column].value_counts(),"\n")
```

```python
le = LabelEncoder()
one_hot = OneHotEncoder()

car['buying'] = le.fit_transform(car['buying'])
car['lug_boot'] = le.fit_transform(car['lug_boot'])
car['maint'] = le.fit_transform(car['maint'])
car['safety'] = le.fit_transform(car['safety'])
car['class'] = le.fit_transform(car['class'])
```

```python
car_enc = pd.get_dummies(car, columns=['doors', 'persons'], dtype=int)
car_enc
```

```python
X = car_enc.drop("class", axis = 1)
X.head()
```

```python
y = car_enc['class']
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, train_size=0.8)
```
