```python
import numpy as np 
import pandas as pd 
import matplotlib 
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.naive_bayes import GaussianNB
from sklearn import mixture
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
le = LabelEncoder()
# tuber['age_label'] = le.fit_transform(tuber['age_years'])
tuber['sex_label'] = le.fit_transform(tuber['sex'])
tuber['residence_type_label'] = le.fit_transform(tuber['residence_type'])
tuber['health_facility_type_label'] = le.fit_transform(tuber['health_facility_type'])
tuber['bcg_vaccination_status_label'] = le.fit_transform(tuber['bcg_vaccination_status'])
tuber['smoking_status_label'] = le.fit_transform(tuber['smoking_status'])
tuber['alcohol_use_label'] = le.fit_transform(tuber['alcohol_use'])
tuber['previous_tb_history_label'] = le.fit_transform(tuber['previous_tb_history'])
tuber['household_tb_contact_label'] = le.fit_transform(tuber['household_tb_contact'])
tuber['malnutrition_status_label'] = le.fit_transform(tuber['malnutrition_status'])
tuber['persistent_cough_label'] = le.fit_transform(tuber['persistent_cough'])
tuber['fever_label']  = le.fit_transform(tuber['fever'])
tuber['night_sweats_label']  = le.fit_transform(tuber['night_sweats'])
tuber['unexplained_weight_loss_label'] = le.fit_transform(tuber['unexplained_weight_loss'])
tuber['chest_pain_label'] = le.fit_transform(tuber['chest_pain'])
tuber['breathlessness_label'] = le.fit_transform(tuber['breathlessness'])
tuber['symptom_count_label'] = le.fit_transform(tuber['symptom_count'])
tuber['sputum_smear_result_label'] = le.fit_transform(tuber['sputum_smear_result'])
tuber['chest_xray_result_label'] = le.fit_transform(tuber['chest_xray_result'])
tuber['gene_xpert_result_label'] = le.fit_transform(tuber['gene_xpert_result'])
tuber['rifampicin_resistance_result_label'] = le.fit_transform(tuber['rifampicin_resistance_result'])
tuber['tuberculosis_status_label'] = le.fit_transform(tuber['tuberculosis_status'])
```

```python
new_da = tuber
```

```python
X = new_da.drop(['patient_id', 'sex', 'state', 'local_government_area','hiv_status','diabetes_status', 'diagnosis_date','tuberculosis_status',
                 'treatment_status','treatment_outcome','age_group', 'age_years', 'residence_type','health_facility_type','bcg_vaccination_status',
                'smoking_status','alcohol_use','previous_tb_history','household_tb_contact','malnutrition_status','fever','night_sweats',
                'unexplained_weight_loss','chest_pain','breathlessness','symptom_count','sputum_smear_result','chest_xray_result','gene_xpert_result',
                'rifampicin_resistance_result','persistent_cough','tuberculosis_status'], axis=1)
```

```python
y = tuber['tuberculosis_status_label']
```

```python
from sklearn.feature_selection import VarianceThreshold
threshold = VarianceThreshold(threshold=.5)
features = threshold.fit_transform(X)
features[:3]
```

```python
Xtrain, Xtest, ytrain, ytest = train_test_split(features, y, random_state=42, train_size=0.8)
```

```python
from sklearn.linear_model import LogisticRegression
reg = LogisticRegression(class_weight='balanced', random_state=42, max_iter=100)
reg.fit(Xtrain,ytrain)
pred = reg.predict(Xtest)
accuracy_score(ytest, pred)
```
